## Contexte

Mise en place d'une chaîne d'observabilité (métriques système et données métier) sur une infrastructure conteneurisée, dans un environnement pédagogique simulant des bases isolées (TAAF). Architecture à deux VM (services applicatifs / monitoring) sur réseau interne. Les adresses, comptes et données correspondent à un lab et ne sont pas reproduits ici.

## Objectifs

- Déployer une stack d'observabilité complète à partir de conteneurs.
- Concevoir des dashboards Grafana (conteneurs + données métier).
- Écrire des requêtes PromQL et SQL pertinentes.
- Mettre en place de l'alerting et de l'analyse de logs.

## Technologies

Docker / Docker Compose · Prometheus · Grafana · Loki · Grafana Alloy · cAdvisor · node_exporter · postgres_exporter · PostgreSQL · PromQL · LogQL · SQL.

## Travaux réalisés

- Validation de l'infrastructure (connectivité inter-VM, pare-feu, réseau Docker, disponibilité des services).
- TP1 — supervision des conteneurs via cAdvisor/Prometheus (état, ressources, I/O disque).
- TP2 — dashboards de données métier interrogeant PostgreSQL (SQL).
- TP3 — intégration de Loki et Alloy pour les logs, définition de règles d'alertes.

## Architecture & fonctionnement

Prometheus collecte les métriques système et conteneurs (cAdvisor + exporters) ; Grafana centralise la visualisation ; Loki agrège les logs collectés par Alloy. Les dashboards combinent métriques techniques (Prometheus) et données métier (PostgreSQL).

## Résultats

Dashboards de supervision fonctionnels (conteneurs et données métier), mécanisme d'alerting et centralisation des logs. Démarche, requêtes et interprétations documentées dans un rapport de TP de 58 pages.

## Difficultés rencontrées

- Communication inter-conteneurs (association aux réseaux Docker, alias DNS).
- Démarrage de Loki avec rétention (configuration du compactor).
- Distinction templating Grafana (`label_values`) vs PromQL / LogQL.

## Compétences développées

Observabilité (métriques, logs, dashboards) · conteneurisation Docker Compose · PromQL / LogQL / SQL · diagnostic réseau · conception d'alertes.

## Améliorations possibles

- Suivi des redémarrages de conteneurs.
- Alertes mémoire proactives.
- Enrichissement des dashboards métier.
