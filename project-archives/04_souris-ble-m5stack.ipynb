## Contexte

Concevoir une interface utilisateur sur M5Stack Core2 (ESP32) qui se comporte comme une souris Bluetooth HID, pilotée par les boutons, l'écran et le capteur d'inclinaison (IMU).

## Objectifs

- Faire fonctionner le M5Stack comme souris Bluetooth HID.
- Déplacer le curseur par l'inclinaison (IMU).
- Définir des zones de clic à l'écran et un mode scroll.
- Ajouter un retour audio (démarrage, connexion).

## Technologies

C++ / Arduino · ESP32 · M5Stack Core2 · arduino-cli · librairie M5Unified (affichage + audio) · librairie BLE Mouse (fork T-vK patché) · capteur IMU · audio WAV converti en `.h` (PROGMEM).

## Travaux réalisés

- Chaîne de compilation (arduino-cli, plateforme M5Stack Core2), flash en ligne de commande.
- Validation du BLE HID (carte reconnue comme souris).
- Lecture IMU (pitch/roll) → déplacement du curseur ; zones d'écran → clics ; bouton central → scroll.
- Intégration audio : WAV → en-têtes `.h` en Flash (PROGMEM).
- Code modulaire : `BleMouseModule`, `ImuDirectionModule`, `ButtonUI`.

## Résultats

Souris Bluetooth opérationnelle : déplacement par inclinaison, clics par zones, scroll, retours audio aux moments clés. Fonctionnalités visées atteintes dans le cadre du TP.

## Difficultés rencontrées

- Librairie BLE Mouse absente de l'index Arduino (installation + patch manuels).
- Contraintes mémoire ESP32 (sons en Flash via PROGMEM).
- Conversion des fichiers audio WAV en en-têtes C.
- Connexion Bluetooth limitée à un appareil à la fois.

## Compétences développées

Programmation embarquée C++/Arduino sur ESP32 · Bluetooth Low Energy (HID) · capteurs inertiels (IMU) · gestion mémoire / intégration de ressources · structuration d'un firmware.
