## Contexte

Projet de challenge autour de l'analyse et de la prédiction des données Vélib' (vélos en libre-service) et météorologiques, avec une architecture Big Data (Apache Spark). Le dépôt est resté à l'état de squelette / cadrage (statut « en développement »).

## Objectifs visés

- Ingestion de données depuis des API publiques (Vélib', météo).
- Traitement massif avec Spark (ETL).
- Modèles ML : prédiction de la demande, clustering de stations, analyse de tendances.
- Visualisation des résultats.

## Technologies prévues

Python · Apache Spark · HDFS · APIs publiques (Vélib', OpenWeatherMap) · Jupyter.

## État réel

Le dépôt contient principalement une structure de projet, un README, un rapport technique partiel et des fichiers de configuration. Les notebooks et résultats n'ont pas été produits ; plusieurs sections restent « à compléter ». À considérer comme un cadrage plutôt qu'un projet abouti.

## Enseignements

Cadrage d'un pipeline Big Data (ingestion → ETL Spark → ML → visualisation) et organisation d'un dépôt de projet data.
