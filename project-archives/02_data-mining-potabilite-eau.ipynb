## Contexte

TP de data mining : construire un modèle de classification binaire par régression logistique, évaluer ses performances et explorer des stratégies d'optimisation. Jeu de données « Water Quality » (Kaggle) : mesures chimiques et biologiques d'échantillons d'eau, cible = potable / non potable (~6 400 échantillons exploitables).

## Objectifs

- Choisir et comprendre un jeu de données à cible binaire.
- Nettoyer les données et mener une analyse exploratoire approfondie.
- Construire des variables de synthèse (feature engineering).
- Entraîner une régression logistique et interpréter la significativité des variables.

## Technologies

Python · pandas · NumPy · scikit-learn · statsmodels · matplotlib / seaborn · Jupyter Notebook.

## Travaux réalisés

- Nettoyage : valeurs `#NUM!` → NaN, gestion des manquants, contrôle des types.
- Analyse exploratoire : distributions, corrélations, boxplots/outliers, pairplots.
- Feature engineering : variables de synthèse (ex. somme des métaux lourds).
- Standardisation et analyse de multicolinéarité (VIF).
- Régression logistique, coefficients et p-values (statsmodels).

## Résultats

Modèle convergent, pseudo-R² ≈ 0,30, variables significatives identifiées (aluminium, arsenic, cadmium, perchlorate…). Déséquilibre de classes marqué pris en compte dans l'interprétation. La sortie statistique complète figure dans le fichier de résultats du projet.

## Difficultés rencontrées

- Nettoyage d'un jeu de données avec erreurs de saisie et valeurs manquantes.
- Gestion de la multicolinéarité (VIF).
- Interprétation rigoureuse dans un contexte déséquilibré.

## Compétences développées

Analyse exploratoire et visualisation · préparation de données · modélisation supervisée et évaluation statistique · écosystème Python data.

## Améliorations possibles

- Rééquilibrage des classes.
- Comparaison avec d'autres modèles (arbres, ensembles).
- Optimisation du seuil de décision, métriques complètes (matrice de confusion, ROC/AUC).
