## Contexte

TP de développement logiciel centré sur la conception orientée objet et la modélisation UML avant implémentation. Cas d'usage : distributeur automatique de produits (sandwichs, boissons, desserts) piloté en ligne de commande.

## Objectifs

- Modéliser un système complet en UML (classes, séquence, activité).
- Implémenter cette modélisation en Java (architecture orientée objet).
- Gérer un état persistant (stock) entre exécutions.
- Simuler un parcours utilisateur réaliste de bout en bout.

## Technologies

Java · programmation orientée objet · UML · machine à états · persistance fichier.

## Fonctionnalités réalisées

- Affichage dynamique du stock, mis à jour en temps réel.
- Commande multi-produits avec regroupement des produits identiques.
- Vérification stricte du stock (pas de sur-commande).
- Paiement espèces (rendu de monnaie) ou carte (simulation terminal).
- Génération d'un ticket, simulation de préparation.
- Persistance du stock dans un fichier texte entre exécutions.

## Architecture & modélisation

Entités : `Produit`, `Commande`, `LigneCommande`, `Paiement`, `SystemePaiement`, `Ticket`, `Distributeur`, `StockManager`. Le distributeur centralise la logique ; une commande regroupe plusieurs lignes ; le paiement est relié au ticket ; le système de paiement est un service externe. Des énumérations gèrent états de commande et modes de paiement. Modélisation UML (classes, séquence, activité) réalisée en amont.

## Résultats

Application fonctionnelle gérant un parcours complet (sélection, paiement, ticket, mise à jour et sauvegarde du stock), appuyée sur une modélisation UML documentée.

## Difficultés rencontrées

- Gestion fiable du stock pendant la commande (cohérence, regroupement).
- Modélisation des états de paiement et de commande.
- Correspondance entre modèle UML et implémentation Java.

## Compétences développées

Conception orientée objet et modélisation UML · développement Java structuré · gestion d'un état persistant · traduction d'une analyse en code.

## Améliorations possibles

- Interface graphique (Swing / JavaFX).
- Historique des commandes et paiements.
- Base de données relationnelle.

## Note

Projet réalisé en équipe dans un cadre pédagogique ; contribution personnelle exacte à préciser.
