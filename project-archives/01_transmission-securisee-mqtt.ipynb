## Contexte

SAE du cursus ingénieur : simuler la transmission de données environnementales (température, humidité…) d'un équipement vers un poste « entreprise » via un broker MQTT, en garantissant la confidentialité des données, y compris vis-à-vis du broker.

## Objectifs

- Comprendre le modèle publish/subscribe de MQTT.
- Implémenter un chiffrement bout-en-bout AES.
- Appliquer une dérivation de clé robuste (PBKDF2) à partir d'un secret partagé.
- Structurer une application Java modulaire et isoler la configuration.

## Technologies

Java 17 · Eclipse Paho MQTT · org.json · Mosquitto (broker) · AES-CBC · PBKDF2-HMAC-SHA256 · SHA-256 · Ubuntu Server/Desktop.

## Architecture & fonctionnement

Trois machines virtuelles Ubuntu sur un même réseau local : émetteur (équipement), récepteur (entreprise) et broker Mosquitto. La clé AES est dérivée d'un identifiant partagé `room` via PBKDF2 ; le topic MQTT est dérivé par SHA-256. L'émetteur chiffre un message JSON et le publie ; le récepteur, abonné au même topic, déchiffre et affiche. Le broker ne manipule que du chiffré. L'adresse du broker est externalisée dans un fichier de configuration.

## Travaux réalisés

- Broker Mosquitto local, validation de la chaîne publish/subscribe.
- Classes Java : `CryptoUtils` (dérivation + chiffrement), `ClientPublisher`, `ClientSubscriber`.
- Chiffrement AES-CBC des messages JSON avant publication.
- Topic dérivé du secret partagé (cloisonnement des échanges).
- Externalisation de la configuration réseau.

## Résultats

Chaîne complète fonctionnelle : message chiffré côté équipement, relayé sans être lisible par le broker, déchiffré et affiché côté entreprise. Objectif de confidentialité bout-en-bout atteint dans le périmètre du projet.

## Difficultés rencontrées

- Dérivation de clé cohérente entre émetteur et récepteur.
- Construction d'un topic sécurisé dérivé du secret.
- Séparation des responsabilités (crypto / réseau / configuration).

## Compétences développées

Cryptographie appliquée (AES, PBKDF2, SHA-256) · protocole MQTT · développement Java structuré · administration Linux et réseau local.

## Améliorations possibles

- Sel aléatoire par message.
- Vérification d'intégrité (HMAC).
- Stockage des données reçues côté entreprise.
- Authentification des clients.
